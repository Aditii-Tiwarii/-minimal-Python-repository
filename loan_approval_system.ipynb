```python
# A loan applicant is represented as a frame containing attributes.
def create_applicant(name, income, credit_score, existing_loan,
                     employment, document_verified):
    return {
        "name": name,
        "income": income,
        "credit_score": credit_score,
        "existing_loan": existing_loan,
        "employment": employment,
        "document_verified": document_verified
    }
```

```python
# Applies rules to the facts stored in the knowledge base.
def make_decision(applicant):
    name = applicant["name"]
    income = applicant["income"]
    credit_score = applicant["credit_score"]
    existing_loan = applicant["existing_loan"]
    employment = applicant["employment"]
    document_verified = applicant["document_verified"]

    reason = []

    # Rule 1: Validate income
    if income is None:
        reason.append("Income information is missing.")
    elif income < 0:
        return "Invalid Data", ["Income cannot be negative."]
    elif income < 20000:
        return "Not Approved", [
            f"Monthly income is Rs. {income}, below the minimum Rs. 20000."
        ]

    # Rule 2: Check credit score
    if credit_score is None:
        reason.append("Credit score is unknown.")
    elif credit_score < 300 or credit_score > 900:
        return "Invalid Data", [
            "Credit score must be between 300 and 900."
        ]
    elif credit_score < 650:
        return "Not Approved", [
            f"Credit score is {credit_score}, below the required 650."
        ]

    # Rule 3: Check existing loan
    if existing_loan is None:
        reason.append("Existing loan status is unknown.")
    elif existing_loan is True:
        reason.append("Applicant already has an existing loan.")

    # Rule 4: Check employment
    if employment is None:
        reason.append("Employment status is unknown.")
    elif employment is False:
        reason.append("Applicant is not currently employed.")

    # Rule 5: Verify documents
    if document_verified is None:
        reason.append("Document verification status is unknown.")
    elif document_verified is False:
        reason.append("Required documents are not verified.")

    # Rule 6: Make the final decision
    if reason:
        return "Conditional Review", reason
    else:
        return "Loan Approved", [
            "Monthly income meets the minimum requirement.",
            "Credit score meets the minimum requirement.",
            "No existing loan was reported.",
            "Applicant is employed.",
            "Required documents are verified."
        ]
```

```python
# Test the knowledge base with different applicants
applicants = [
    create_applicant("Aman", 50000, 780, False, True, True),
    create_applicant("Simran", 15000, 720, False, True, True),
    create_applicant("Ravi", 40000, 580, False, True, True),
    create_applicant("Neha", 45000, 750, None, True, True),
    create_applicant("Karan", 60000, 800, False, False, True),
    create_applicant("Priya", 35000, 700, False, True, False)
]
```

```python
# Display decisions and explanations
for applicant in applicants:
    decision, reasons = make_decision(applicant)

    print("\nApplicant:", applicant["name"])
    print("Loan Decision:", decision)
    print("Reasoning:")

    for reason in reasons:
        print("-", reason)
```
